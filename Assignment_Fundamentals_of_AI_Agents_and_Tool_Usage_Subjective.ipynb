```python
import pandas as pd


# ----------------------------------------------------------
# Tool 1: List all column names
# ----------------------------------------------------------
def list_columns(df):
    """
    Returns the list of column names in the DataFrame.
    """
    return list(df.columns)


# ----------------------------------------------------------
# Tool 2: Get statistics for a numeric column
# ----------------------------------------------------------
def get_stats(df, column):
    """
    Returns count, mean, max, min, and sum
    for the specified numeric column.
    """
    if column not in df.columns:
        return {"error": f"Column '{column}' not found"}

    if not pd.api.types.is_numeric_dtype(df[column]):
        return {"error": f"Column '{column}' is not numeric"}

    return {
        "count": int(df[column].count()),
        "mean": float(df[column].mean()),
        "max": float(df[column].max()),
        "min": float(df[column].min()),
        "sum": float(df[column].sum())
    }


# ----------------------------------------------------------
# Tool 3: Filter rows and return count and names
# ----------------------------------------------------------
def filter_count(df, column, value):
    """
    Returns the number of matching rows,
    the matching names,
    and the filtered DataFrame.
    """
    if column not in df.columns:
        return {"error": f"Column '{column}' not found"}

    filtered_df = df[df[column] == value]

    return {
        "count": len(filtered_df),
        "names": filtered_df["Name"].tolist(),
        "filtered_df": filtered_df
    }


# ----------------------------------------------------------
# Dispatcher (similar to available_tools pattern)
# ----------------------------------------------------------
def call_tool(tool_name, arguments, df):
    """
    Dispatches a tool call using a dictionary
    of available tools.
    """

    available_tools = {
        "list_columns": list_columns,
        "get_stats": get_stats,
        "filter_count": filter_count
    }

    if tool_name not in available_tools:
        raise ValueError(f"Unknown tool: {tool_name}")

    tool = available_tools[tool_name]

    return tool(df, **arguments)


# ----------------------------------------------------------
# Main Program
# ----------------------------------------------------------
if __name__ == "__main__":

    employee_data = {
        "Name": [
            "Rahul", "Lakshmi", "Manoj", "Divya",
            "Vikram", "Anita", "Karthik", "Deepa",
            "Suresh", "Kavya", "Sanjay", "Arun"
        ],
        "Department": [
            "Engineering", "Marketing", "Engineering", "Marketing",
            "Engineering", "Sales", "Marketing", "Engineering",
            "Sales", "Marketing", "Engineering", "Engineering"
        ],
        "Salary": [
            75500, 62000, 81000, 58000,
            79500, 55000, 60500, 84000,
            57000, 61000, 76000, 78000
        ],
        "Experience (years)": [
            5, 4, 7, 3,
            6, 4, 5, 8,
            3, 4, 5, 6
        ],
        "City": [
            "Hyderabad", "Pune", "Bangalore", "Delhi",
            "Hyderabad", "Mumbai", "Chennai", "Hyderabad",
            "Pune", "Bangalore", "Delhi", "Hyderabad"
        ]
    }

    # Create DataFrame
    df = pd.DataFrame(employee_data)

    # Display column names
    print("Columns:")
    print(call_tool("list_columns", {}, df))

    # Step 1: Filter Engineering employees
    result = call_tool(
        "filter_count",
        {
            "column": "Department",
            "value": "Engineering"
        },
        df
    )

    engineering_df = result["filtered_df"]

    # Step 2: Get salary statistics for Engineering employees
    salary_stats = call_tool(
        "get_stats",
        {
            "column": "Salary"
        },
        engineering_df
    )

    # Step 3: Print required output
    print("\nEngineering Employee Analysis")
    print("-" * 40)
    print("Engineering count =", result["count"])
    print("Names =", result["names"])
    print("Average salary =", salary_stats["mean"])
```
