```python
#Task-1
def calculate_total(marks):
    """Return the sum of all marks in the list."""
    total = 0
    for mark in marks:
        total += mark
    return total
```

```python
#Task-2

def calculate_average(marks):
    """Return the average of marks using calculate_total()."""
    total = calculate_total(marks)
    return total / len(marks)
```

```python
#Task-3
def get_grade(average):
    """Return grade based on average score."""
    if average > 90:
        return "A"
    elif average > 75:
        return "B"
    else:
        return "C"
```

```python
#Task-4
def display_report(marks):
    """Display total, average, and grade for the given marks."""
    total = calculate_total(marks)
    average = calculate_average(marks)
    grade = get_grade(average)

    print("Total:", total)
    print("Average:", average)
    print("Grade:", grade)
```

```python
def find_min(numbers):
    """ Returns the smallest number in a list.
    Args:
        numbers (list): A list of numeric values.
    Returns:
        int/float: The smallest value in the list.
    """
    return min(numbers)
```

```python
#Task-1
def calculate_total(marks):
    """Return the sum of all marks in the list."""
    return sum(marks)

#Task-2
def calculate_average(marks):
    """Return the average of the marks by using calculate_total."""
    total = calculate_total(marks)
    return total / len(marks) if marks else 0

#Task-3
def get_grade(average):
    """Return grade based on the average score."""
    if average > 90:
        return "A"
    elif average > 75:
        return "B"
    else:
        return "C"

#Task-4
def display_report(marks):
    """Display total, average, and grade using the other functions."""
    total = calculate_total(marks)
    average = calculate_average(marks)
    grade = get_grade(average)

    print(f"Total: {total}")
    print(f"Average: {average}")
    print(f"Grade: {grade}")


# Task-5
marks_list = [88, 76, 95, 60, 82]
display_report(marks_list)
```
