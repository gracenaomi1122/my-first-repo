```python
# -------------------------------
# Sample Input
# -------------------------------

system_persona = "You are an expert career coach."

turns = [
    "What is the roadmap to become an FDE in 6 months?",
    "Summarize your previous answer in one sentence."
]


# --------------------------------------------------------
# Mock model function
# --------------------------------------------------------
# This function simulates a real LLM response.
# Instead of making an API request, it returns a predictable
# (deterministic) response based on the current conversation.
def mock_call_model(messages):
    """
    Simulates an LLM response.

    Parameters:
        messages (list): Current conversation history.

    Returns:
        str: A mock assistant reply.
    """

    # Count how many assistant replies already exist
    reply_number = sum(
        1 for message in messages
        if message["role"] == "assistant"
    ) + 1

    # Return a deterministic response
    return (
        f"Mock reply #{reply_number} "
        f"based on {len(messages)} messages so far."
    )


# --------------------------------------------------------
# Function to simulate a conversation
# --------------------------------------------------------
def run_conversation(system_persona, turns, call_model):
    """
    Builds and maintains the complete conversation history.

    Steps:
    1. Add the system persona.
    2. For every user turn:
         a. Add the user's message.
         b. Call the model using the current conversation.
         c. Store the assistant's reply.
    3. Return the complete conversation history.

    Parameters
    ----------
    system_persona : str
        Instructions that define the assistant's behavior.

    turns : list
        List of user messages.

    call_model : function
        Function that receives the message history and
        returns an assistant reply.

    Returns
    -------
    list
        Complete conversation history.
    """

    # --------------------------------------------
    # Step 1: Create the conversation history
    # --------------------------------------------
    messages = []

    # Add the system message first
    messages.append({
        "role": "system",
        "content": system_persona
    })

    # --------------------------------------------
    # Step 2: Process every user turn
    # --------------------------------------------
    for turn_number, user_message in enumerate(turns, start=1):

        # Add the user's message
        messages.append({
            "role": "user",
            "content": user_message
        })

        # Call the model with the current history
        assistant_reply = call_model(messages)

        # Store the assistant's reply
        messages.append({
            "role": "assistant",
            "content": assistant_reply
        })

    # --------------------------------------------
    # Step 3: Return the complete history
    # --------------------------------------------
    return messages


# --------------------------------------------------------
# Driver Code
# --------------------------------------------------------
if __name__ == "__main__":

    # Run the conversation
    final_messages = run_conversation(
        system_persona,
        turns,
        mock_call_model
    )

    # Print the final conversation history
    print("Final Conversation History:\n")

    for index, message in enumerate(final_messages, start=1):
        print(f"Message {index}")
        print(f"Role    : {message['role']}")
        print(f"Content : {message['content']}")
        print("-" * 60)

    # Print the returned list
    print("\nReturned Messages List:")
    print(final_messages)
```

```python
system_persona = "You are an expert career coach."

turns = [
    "What is the roadmap to become an FDE in 6 months?",
    "Summarize your previous answer in one sentence."
]

def mock_call_model(messages):
    # Deterministic stand-in for a real LLM API call, used only for local testing.
    reply_number = sum(1 for m in messages if m["role"] == "assistant") + 1
    return f"Mock reply #{reply_number} based on {len(messages)} messages so far."


def run_conversation(system_persona, turns, call_model):
    # Start the conversation with the system persona
    messages = [
        {"role": "system", "content": system_persona}
    ]

    # Process each user turn
    for turn in turns:
        # Append the user's message
        messages.append({"role": "user", "content": turn})

        # Get the assistant's reply using the stand-in model
        reply = call_model(messages)

        # Append the assistant's reply
        messages.append({"role": "assistant", "content": reply})

    return messages


if __name__ == "__main__":
    result = run_conversation(system_persona, turns, mock_call_model)
    print(result)
```
