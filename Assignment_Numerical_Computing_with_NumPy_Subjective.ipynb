```python
import numpy as np

#Task 1
#Generate and Inspect the Dataset

#Set seed
np.random.seed(42)

#List of names
names_list = ["Alice", "Bob", "Charlie", "David", "Emma"]

#Generate data for 10 students
names = np.random.choice(names_list, 10)
ages = np.random.randint(15, 19, 10)

maths = np.random.randint(40, 101, 10)
science = np.random.randint(40, 101, 10)
english = np.random.randint(40, 101, 10)

#Creating marks array
marks = np.column_stack((maths, science, english))

#Print shape and dtype
print("Marks Array:\n", marks)
print("\nShape of marks array:", marks.shape)
print("Data type of marks array:", marks.dtype)

#Task 2 — Reshape and Slice the Marks Array

#Reshape array
reshaped_marks = marks.reshape(2, 5, 3)

print("\nReshaped Array (2, 5, 3):\n", reshaped_marks)

#Slice first five students for maths and science only
slice_marks = marks[:5, 0:2]

print("\nFirst five students (Maths & Science only):\n", slice_marks)


#Task 3 — Save Records to a File

with open("student_records.txt", "w") as file:
    for i in range(10):
        record = f"{names[i]} | {ages[i]} | {maths[i]} | {science[i]} | {english[i]}\n"
        file.write(record)

print("\nRecords successfully written to student_records.txt")

#Read and print file contents
print("\nContents of student_records.txt:\n")

with open("student_records.txt", "r") as file:
    lines = file.readlines()
    for line in lines:
        print(line.strip())
```

```python
import numpy as np

np.random.seed(42)

# ── Task 1: Generate and Inspect the Dataset ──────────────────────────────────
names   = np.random.choice(['Alice', 'Bob', 'Carol', 'David', 'Eva'], size=10)
ages    = np.random.randint(15, 19, size=10)
maths   = np.random.randint(40, 101, size=10)
science = np.random.randint(40, 101, size=10)
english = np.random.randint(40, 101, size=10)

marks = np.array([maths, science, english]).T   # shape (10, 3)

print("Marks array shape :", marks.shape)
print("Marks array dtype :", marks.dtype)

# ── Task 2: Reshape and Slice ─────────────────────────────────────────────────
reshaped = marks.reshape(2, 5, 3)
print("\nReshaped marks array (2, 5, 3):")
print(reshaped)

sliced = marks[:5, :2]
print("\nFirst 5 students — Maths and Science only:")
print(sliced)

# Shape (2, 5, 3) means the 10 student records are divided into 2 groups,
# each group containing 5 students, and each student has 3 subject marks.
# It is like splitting the class into 2 batches of 5 students each.

# ── Task 3: Save Records to File ─────────────────────────────────────────────
with open('student_records.txt', 'w') as f:
    for i in range(10):
        line = f"{names[i]} | {ages[i]} | {maths[i]} | {science[i]} | {english[i]}\n"
        f.write(line)

print("\nRecords written to student_records.txt")

with open('student_records.txt', 'r') as f:
    lines = f.readlines()

print("\nVerification — File Contents:")
for line in lines:
    print(line.strip())
```
