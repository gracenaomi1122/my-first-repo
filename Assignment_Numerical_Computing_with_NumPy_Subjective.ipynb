```python
import numpy as np

#Task 1
#Generate and Inspect the Dataset

#Set seed
np.random.seed(42)

#List of names
names_list = ["Alice", "Bob", "Charlie", "David", "Emma"]

#Generate data for 10 students
names = np.random.choice(names_list, 10)
ages = np.random.randint(15, 19, 10)

maths = np.random.randint(40, 101, 10)
science = np.random.randint(40, 101, 10)
english = np.random.randint(40, 101, 10)

#Creating marks array
marks = np.column_stack((maths, science, english))

#Print shape and dtype
print("Marks Array:\n", marks)
print("\nShape of marks array:", marks.shape)
print("Data type of marks array:", marks.dtype)

#Task 2 — Reshape and Slice the Marks Array

#Reshape array
reshaped_marks = marks.reshape(2, 5, 3)

print("\nReshaped Array (2, 5, 3):\n", reshaped_marks)

#Slice first five students for maths and science only
slice_marks = marks[:5, 0:2]

print("\nFirst five students (Maths & Science only):\n", slice_marks)


#Task 3 — Save Records to a File

with open("student_records.txt", "w") as file:
    for i in range(10):
        record = f"{names[i]} | {ages[i]} | {maths[i]} | {science[i]} | {english[i]}\n"
        file.write(record)

print("\nRecords successfully written to student_records.txt")

#Read and print file contents
print("\nContents of student_records.txt:\n")

with open("student_records.txt", "r") as file:
    lines = file.readlines()
    for line in lines:
        print(line.strip())
```
