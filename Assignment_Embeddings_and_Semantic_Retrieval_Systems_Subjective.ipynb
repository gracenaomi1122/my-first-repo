```python
# ==========================================================
# Semantic Search using Cosine Similarity
# ==========================================================

# Import NumPy library
# NumPy provides mathematical operations on vectors.
import numpy as np

# ==========================================================
# Step 1: Knowledge Base
# ==========================================================
# Each sentence represents a document in the knowledge base.

knowledge_base = [
    "Linear regression finds the best fit line through the data points.",
    "Decision trees make predictions by asking a series of yes/no questions.",
    "Random forest combines hundreds of decision trees and lets them vote, reducing overfitting and improving accuracy.",
    "Overfitting happens when a model memorizes the training data, including noise."
]

# ==========================================================
# Step 2: Precomputed Embedding Vectors
# ==========================================================
# Each sentence has already been converted into a 4-dimensional
# embedding vector.

kb_embeddings = [

    # Index 0
    [0.10, 0.85, -0.05, 0.20],

    # Index 1
    [0.30, 0.10, 0.60, -0.15],

    # Index 2
    [0.28, 0.15, 0.58, -0.10],

    # Index 3
    [0.05, -0.20, 0.10, 0.90]
]

# ==========================================================
# Step 3: Query Embedding
# ==========================================================
# The user's question has already been converted into
# an embedding vector.

query_embedding = [0.06, -0.18, 0.12, 0.88]


# ==========================================================
# Step 4: Cosine Similarity Function
# ==========================================================

def cosine_similarity(vector_1, vector_2):
    """
    Computes cosine similarity between two vectors.

    Formula:

                 A · B
    --------------------------------
    |A| × |B|

    where

    A · B = Dot Product
    |A|   = Magnitude of vector A
    |B|   = Magnitude of vector B
    """

    # Convert lists into NumPy arrays
    vector_1 = np.array(vector_1)
    vector_2 = np.array(vector_2)

    # ------------------------------------------
    # Compute Dot Product
    # ------------------------------------------
    dot_product = np.dot(vector_1, vector_2)

    # ------------------------------------------
    # Compute Magnitudes
    # ------------------------------------------
    magnitude_vector1 = np.linalg.norm(vector_1)
    magnitude_vector2 = np.linalg.norm(vector_2)

    # ------------------------------------------
    # Compute Cosine Similarity
    # ------------------------------------------
    similarity = dot_product / (
        magnitude_vector1 * magnitude_vector2
    )

    return similarity


# ==========================================================
# Step 5: Semantic Search Function
# ==========================================================

def semantic_search(
        query_embedding,
        kb_embeddings,
        knowledge_base,
        top_k=2
):
    """
    Performs semantic search.

    Parameters
    ----------
    query_embedding : list
        Embedding vector of the user query.

    kb_embeddings : list
        Embedding vectors of knowledge-base entries.

    knowledge_base : list
        Text documents.

    top_k : int
        Number of top matches to return.

    Returns
    -------
    List of tuples:
    (similarity_score, index, text)
    """

    # Store all search results
    results = []

    # Compare query with every document
    for index in range(len(kb_embeddings)):

        # Current document embedding
        current_embedding = kb_embeddings[index]

        # Compute similarity score
        similarity_score = cosine_similarity(
            query_embedding,
            current_embedding
        )

        # Save result
        results.append(
            (
                round(similarity_score, 4),
                index,
                knowledge_base[index]
            )
        )

    # ------------------------------------------
    # Sort by similarity (highest first)
    # ------------------------------------------
    results.sort(
        key=lambda item: item[0],
        reverse=True
    )

    # Return Top-K matches
    return results[:top_k]


# ==========================================================
# Step 6: Main Program
# ==========================================================

if __name__ == "__main__":

    print("=" * 70)
    print("SEMANTIC SEARCH")
    print("=" * 70)

    # Find the top 2 most similar documents
    top_results = semantic_search(
        query_embedding=query_embedding,
        kb_embeddings=kb_embeddings,
        knowledge_base=knowledge_base,
        top_k=2
    )

    # Display results
    print("\nTop Matching Documents\n")

    for score, index, text in top_results:

        print(f"Document Index : {index}")
        print(f"Similarity     : {score}")
        print(f"Document Text  : {text}")
        print("-" * 70)
```
