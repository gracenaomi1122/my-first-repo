```python
import json

# ---------------- Task 1 ----------------
# Create and serialize a nested JSON object

student = {
    "name": "Rahul",
    "age": 21,
    "passed": True,
    "marks": [85, 90, 78],
    "address": {
        "city": "Mumbai",
        "pincode": "400001"
    }
}

# Convert to formatted JSON string
json_string = json.dumps(student, indent=4)
print("Task 1 Output:")
print(json_string)


# ---------------- Task 2 ----------------
# Parse JSON and access values

parsed_data = json.loads(json_string)

print("\nTask 2 Output:")
print("Name:", parsed_data["name"])
print("Second Mark:", parsed_data["marks"][1])
print("City:", parsed_data["address"]["city"])


# ---------------- Task 3 ----------------
# Identify and fix invalid JSON

# Errors in given JSON:
# 1. Keys must be in double quotes (name is not quoted)
# 2. true should be lowercase in JSON (this part is correct actually, but Python string must follow JSON format strictly)

bad_json = '{ "name": "Raji", "score": 88, "active": true }'

# Parse fixed JSON
fixed_data = json.loads(bad_json)

print("\nTask 3 Output:")
print(fixed_data)


# ---------------- Task 4 ----------------
# Build a JSON array of objects

products = [
    {"product_name": "Laptop", "price": 75000, "in_stock": True},
    {"product_name": "Mouse", "price": 500, "in_stock": True},
    {"product_name": "Keyboard", "price": 1500, "in_stock": False}
]

products_json = json.dumps(products, indent=2)

print("\nTask 4 Output:")
print(products_json)
```
