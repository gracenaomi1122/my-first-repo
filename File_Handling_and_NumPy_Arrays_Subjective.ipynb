```python
# Task 1: Create and write to file
with open("warehouse_data.txt", "w") as file:
    values = [120, 85, 200, 95, 150]
    for v in values:
        file.write(str(v) + "\n")
```

```python
# Task 2: Read the file
with open("warehouse_data.txt", "r") as file:
    lines = file.readlines()
    cleaned_values = [line.strip() for line in lines]

# Print each value
for val in cleaned_values:
    print(val)
```

```python
#Task-3

import numpy as np

# Convert to integers
int_values = [int(v) for v in cleaned_values]

# Create NumPy array
arr = np.array(int_values)

# Print properties
print("Array:", arr)
print("Data type:", arr.dtype)
print("Shape:", arr.shape)
print("Size:", arr.size)

# Compute statistics
print("Sum:", np.sum(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
```
