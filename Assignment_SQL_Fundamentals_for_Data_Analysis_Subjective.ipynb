```python
import sqlite3
import pandas as pd

# Create database connection
connection = sqlite3.connect('ecommerce.db')

# Enable foreign key support in SQLite
connection.execute("PRAGMA foreign_keys = ON")

cursor = connection.cursor()

# Create tables with proper constraints
# Customers table
cursor.execute("""
CREATE TABLE IF NOT EXISTS customers (
    customer_id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    city TEXT NOT NULL,
    UNIQUE(name, city)
)
""")

# Orders table
cursor.execute("""
CREATE TABLE IF NOT EXISTS orders (
    order_id INTEGER PRIMARY KEY AUTOINCREMENT,
    customer_id INTEGER NOT NULL,
    amount REAL NOT NULL,
    order_date DATE NOT NULL,
    FOREIGN KEY (customer_id) REFERENCES customers(customer_id)
)
""")

# Insert sample customer data using executemany()

customers_data = [
    ("John Smith", "New York"),
    ("Alice Johnson", "Los Angeles"),
    ("Michael Brown", "New York")
]

cursor.executemany("""
INSERT OR IGNORE INTO customers (name, city)
VALUES (?, ?)
""", customers_data)

# Commit customer inserts before fetching IDs
connection.commit()
# Fetch customer IDs for inserting orders

cursor.execute("SELECT customer_id, name FROM customers")
customer_records = cursor.fetchall()

# Create dictionary: {customer_name: customer_id}
customer_dict = {name: customer_id for customer_id, name in customer_records}

# Insert sample order data using executemany()

orders_data = [
    (customer_dict["John Smith"], 150.75, "2026-05-20"),
    (customer_dict["John Smith"], 85.50, "2026-05-21"),
    (customer_dict["Alice Johnson"], 200.00, "2026-05-22"),
    (customer_dict["Michael Brown"], 120.00, "2026-05-23")
]

cursor.executemany("""
INSERT INTO orders (customer_id, amount, order_date)
VALUES (?, ?, ?)
""", orders_data)

# Commit order inserts
connection.commit()

# Query:
# Retrieve customers from New York with orders > 100

query = """
SELECT
    customers.name,
    customers.city,
    orders.amount
FROM customers
INNER JOIN orders
    ON customers.customer_id = orders.customer_id
WHERE customers.city = 'New York'
AND orders.amount > 100
"""

# Load query results into a Pandas DataFrame
df_results = pd.read_sql_query(query, connection)
# Close the database connection


connection.close()

# Display results

if __name__ == "__main__":
    print("Customers from New York with orders over $100:\n")
    print(df_results)
```

```python
#Masai Solution


import sqlite3

import pandas as pd

# Create database connection

connection = sqlite3.connect('ecommerce.db')

cursor = connection.cursor()

# Create customers table

cursor.execute('''

  CREATE TABLE IF NOT EXISTS customers (

    customer_id INTEGER PRIMARY KEY AUTOINCREMENT,

    name TEXT NOT NULL,

    city TEXT NOT NULL

  )

''')

# Create orders table

cursor.execute('''

  CREATE TABLE IF NOT EXISTS orders (

    order_id INTEGER PRIMARY KEY AUTOINCREMENT,

    customer_id INTEGER NOT NULL,

    amount REAL NOT NULL,

    order_date DATE NOT NULL,

    FOREIGN KEY (customer_id) REFERENCES customers(customer_id)

  )

''')

# Clear existing data for clean test runs

cursor.execute('DELETE FROM orders')

cursor.execute('DELETE FROM customers')

# Insert sample customer data

customers_data = [

  ('Alice Johnson', 'New York'),

  ('Bob Smith', 'Boston'),

  ('Charlie Brown', 'New York')

]

cursor.executemany('''

  INSERT INTO customers (name, city)

  VALUES (?, ?)

''', customers_data)

# Insert sample order data

orders_data = [

  (1, 150.00, '2024-01-10'),

  (1, 75.00, '2024-01-15'),

  (2, 200.00, '2024-01-12'),

  (3, 110.00, '2024-01-18')

]

cursor.executemany('''

  INSERT INTO orders (customer_id, amount, order_date)

  VALUES (?, ?, ?)

''', orders_data)

# Commit changes

connection.commit()

# Query: Get customers from New York with orders over 100

query = '''

SELECT c.name, c.city, o.amount

FROM customers c

INNER JOIN orders o ON c.customer_id = o.customer_id

WHERE c.city = 'New York' AND o.amount > 100

ORDER BY o.amount DESC

'''

# Load results into Pandas DataFrame

df_results = pd.read_sql(query, connection)

# Close connection

connection.close()

# Display results

if __name__ == "__main__":

  print(df_results)

```
