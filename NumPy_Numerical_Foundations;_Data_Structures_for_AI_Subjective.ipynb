```python
#Task-1

# Create a tuple of five city names
cities = ("New York", "London", "Tokyo", "Paris", "Berlin")

# Unpack it into five separate variables
city1, city2, city3, city4, city5 = cities
print(f"Cities: {city1}, {city2}, {city3}, {city4}, {city5}")

# Try to modify the first element and handle the error
try:
    cities[0] = "Amsterdam"
except TypeError:
    print("\nError: Tuples are immutable and cannot be modified after creation.")
```

```python
#Task-2
python_passers = {"Alice", "Bob", "Charlie", "David"}
sql_passers = {"Charlie", "David", "Eve", "Frank"}

# Union: Students who passed either test
print("Union (All passers):", python_passers | sql_passers)

# Intersection: Students who passed both tests
print("Intersection (Passed both):", python_passers & sql_passers)

# Difference: Students who passed Python but NOT SQL
print("Difference (Python only):", python_passers - sql_passers)
```

```python
#Task -3

# List of five integers representing ASCII values
ascii_values = [65, 72, 105, 33, 50]

# Loop with chr() to print integer and its character
print("ASCII Mappings:")
for val in ascii_values:
    print(f"Value {val} corresponds to character: {chr(val)}")

# Use ord() to retrieve the ASCII value of 'Z'
z_value = ord('Z')
print(f"\nThe ASCII value of 'Z' is: {z_value}")
```
