```python
import pandas as pd

# Load the Titanic dataset
df = pd.read_csv("titanic.csv")

# Filter female passengers in pclass 1 or 2
filtered_df = df[
    (df["sex"] == "female") &
    (df["pclass"].isin([1, 2]))
]

# Fill missing age values with 30
filtered_df["age"] = filtered_df["age"].fillna(30)

# Sort by fare in descending order
filtered_df = filtered_df.sort_values(
    by="fare",
    ascending=False
)

# Reset the index
filtered_df = filtered_df.reset_index(drop=True)

# Add a new column with constant value
filtered_df["fare_group"] = "high"

# Save the final DataFrame to CSV
filtered_df.to_csv("female_upper_class.csv", index=False)

print("Processing complete. File saved as female_upper_class.csv")
```

Walkthrough
1. Load the Dataset
df = pd.read_csv("titanic.csv")
This reads the Titanic CSV file into a Pandas DataFrame called df.

2. Filter the Required Rows
filtered_df = df[
    (df["sex"] == "female") &
    (df["pclass"].isin([1, 2]))
]
This keeps only rows where:
•	sex is "female"
•	pclass is either 1 or 2
The conditions are wrapped in parentheses because they are combined with &, which is required in Pandas boolean filtering.

3. Fill Missing Age Values
filtered_df["age"] = filtered_df["age"].fillna(30)
The Titanic dataset contains missing values in the age column. This replaces all missing ages with 30.

4. Sort by Fare
filtered_df = filtered_df.sort_values(
    by="fare",
    ascending=False
)
Rows are sorted so passengers with the highest fares appear first.
5. Reset the Index
filtered_df = filtered_df.reset_index(drop=True)
After sorting, the original row indices remain. reset_index(drop=True) creates clean sequential indices starting from 0.

6. Add a New Column
filtered_df["fare_group"] = "high"
A new column named fare_group is added, and every row gets the constant value "high".

7. Save to CSV
filtered_df.to_csv("female_upper_class.csv", index=False)
The final processed dataset is exported to a new CSV file without including the DataFrame index.

Trade-offs & Limitations
Strengths
•	Simple and readable
•	Uses only the required Pandas operations
•	Efficient for datasets of Titanic size
•	Handles missing ages safely with fillna
Limitations
•	Assumes the CSV file exists and contains the required columns
•	No error handling for missing files or malformed data
•	Assigning directly to filtered_df["age"] may produce a Pandas warning in some environments (SettingWithCopyWarning)
•	For extremely large datasets, loading the entire CSV into memory may become inefficient
