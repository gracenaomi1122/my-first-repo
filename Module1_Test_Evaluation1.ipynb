```python
import numpy as np

# Task 1: even numbers from 2 to 20 inclusive
evens = np.arange(2, 22, 2)
print('evens:', evens)

# Task 2: 9 evenly spaced values between 0.1 and 0.9
tenths = np.linspace(0.1, 0.9, 9)
print('tenths:', tenths)

# Task 3: 3x4 array filled with 7
grid = np.full((3, 4), 7)
print('grid:\n', grid)

# Task 4: element-wise squaring
squared = evens ** 2
print('squared:', squared)

# Task 5: shape and ndim of grid
print(grid.shape)
print(grid.ndim)
```
