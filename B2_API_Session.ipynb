```python
from google.colab import userdata
chat_gpt_key = userdata.get('chat_gpt_key')
print(chat_gpt_key)
```

```python
import requests

url = "https://api.openai.com/v1/chat/completions"

headers = {
    "Authorization": f"Bearer {chat_gpt_key}",
    "Content-Type": "application/json",
}

body = {
    "model": "gpt-4o-mini",
    "messages": [
        {"role": "system", "content": "You are a helpful Python tutor. Keep replies under 3 sentences."},
        {"role": "user", "content": "How to use Google Colab?"},
    ],
}

response = requests.post(url, headers=headers, json=body)

print(response)
```

```python
print("Reply:", response.json()["choices"][0]["message"]["content"])
```
