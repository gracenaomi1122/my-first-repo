```python
# Model Selection & Cost Decision Memo

## Executive Summary

For the launch of our new production customer-support chatbot, we recommend selecting **Claude Opus**. While Kimi K3 presents a lower baseline cost, roughly 95% of companies building production-level AI agents rely on trusted, branded providers like Anthropic (Claude) or OpenAI due to critical requirements around system reliability and trust. Claude Opus offers superior context capacity (300,000 tokens versus GPT-5.6's 272,000 tokens) at a lower monthly operational cost than GPT-5.6 ($70 at launch; $700 at 10x scale), delivering the optimal balance of enterprise trust, context flexibility, and long-term cost efficiency.

---

## Cost Comparison

### Monthly Volume Assumptions

* **Launch Volume:** 4 million input tokens + 2 million output tokens
* **10x-Scaled Volume:** 40 million input tokens + 20 million output tokens

### Cost Breakdown Table

| Model | Launch Monthly Cost | 10x-Scale Monthly Cost |
| --- | --- | --- |
| **Kimi K3** | **$42** | **$420** |
| **GPT-5.6 (top-tier)** | **$80** | **$800** |
| **Claude Opus (flagship)** | **$70** | **$700** |

*Methodology:*

* **Kimi K3:** Launch = (4 × $3) + (2 × $15) = $12 + $30 = **$42** | 10x Scale = (40 × $3) + (20 × $15) = $120 + $300 = **$420**
* **GPT-5.6:** Launch = (4 × $5) + (2 × $30) = $20 + $60 = **$80** | 10x Scale = (40 × $5) + (20 × $30) = $200 + $600 = **$800**
* **Claude Opus:** Launch = (4 × $5) + (2 × $25) = $20 + $50 = **$70** | 10x Scale = (40 × $5) + (20 × $25) = $200 + $500 = **$700**

---

## Production-Readiness Assessment

* **Reliability & Trust Factor:** Enterprise production deployment requires predictable uptime and data integrity. Roughly 95% of companies building production-level AI agents choose trusted, branded providers—primarily Claude or OpenAI—despite higher pricing, because reliability and brand trust outweigh initial cost savings for user-facing support.
* **Open-Source & Chinese-Origin Adoption Risk:** Very few production engineering teams deploy open-source models, and almost none deploy Chinese-origin models (such as Kimi K3 or DeepSeek) for production customer support. Despite strong cost-to-performance ratios, these models introduce long-term reliability concerns and severe data-related constraints that risk compliance and operational continuity.

---

## Latency & Infrastructure

* **API Access vs. On-Premise Hosting:** Accessing models via cloud APIs (e.g., Anthropic or OpenAI) delivers fast response times returned in seconds or milliseconds. Conversely, on-premise hosting of large open-source Mixture-of-Experts (MoE) models (such as Llama 4 Scout) for ~1,000 concurrent users requires roughly 8 high-end GPUs, driving monthly infrastructure costs to $27,000–$45,000 plus substantial developer maintenance overhead.
* **Hardware Bottlenecks:** Self-hosted infrastructure is hard-constrained by internal GPU throughput. Insufficient compute resources or power supply limits can force a self-hosted platform to pause new user onboarding, making API consumption significantly more scalable and cost-effective for our initial workload.

---

## Token Caching Strategy

* **Cost Reduction Lever:** Providers offer reduced per-token rates for cached input tokens compared to fresh input tokens, providing a clear financial mechanism to optimize repetitive prompt structures, static system prompts, and core documentation contexts.
* **Trade-Off & Quality Risk:** Caching stale context data can lead to degraded answer quality or outdated information being served to customers. Furthermore, toggling very large context configurations (such as a maximum 1-million-token mode) roughly doubles per-token API pricing, meaning caching policies must balance prompt update frequencies against context size costs.

---

## Final Recommendation

* **Chosen Model:** **Claude Opus** is recommended as the production LLM power source, providing proven tier-one reliability, a high context window (300,000 tokens), and a lower monthly operational cost than GPT-5.6 ($70/month at launch scaling to $700/month at 10x volume).
* **Post-Launch Monitoring Metrics:**
1. **Average API Response Latency:** Track overall response times (in milliseconds/seconds) to detect provider-side degradation during peak chat volume.
2. **Cache Hit Ratio & Cache Staleness Rate:** Monitor the percentage of input tokens served via cache versus fresh inputs to track cost optimization without sacrificing customer support answer freshness.
```
