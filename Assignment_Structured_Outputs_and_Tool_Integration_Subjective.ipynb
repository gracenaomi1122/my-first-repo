```python
# ==========================================================
# Idempotent Order Processing Tool
# ==========================================================

processed_orders = {}


def process_order_requests(requests):
    """
    Processes order requests with:
    1. Quantity validation
    2. Idempotency using request_id
    """

    results = []

    for req in requests:

        request_id = req["request_id"]
        quantity = req["quantity"]

        # -----------------------------
        # Validate quantity
        # -----------------------------
        if not isinstance(quantity, int) or quantity <= 0:
            results.append({
                "request_id": request_id,
                "status": "error",
                "message": "Quantity must be an integer greater than 0.",
                "order": None
            })
            continue

        # -----------------------------
        # Duplicate request check
        # -----------------------------
        if request_id in processed_orders:
            results.append({
                "request_id": request_id,
                "status": "success",
                "message": "Request already processed (duplicate tool call).",
                "order": processed_orders[request_id]
            })
            continue

        # -----------------------------
        # Create new order
        # -----------------------------
        order = {
            "request_id": request_id,
            "product": req["product"],
            "quantity": quantity
        }

        processed_orders[request_id] = order

        results.append({
            "request_id": request_id,
            "status": "success",
            "message": "Order processed successfully.",
            "order": order
        })

    return results


# ==========================================================
# Main Program
# ==========================================================
if __name__ == "__main__":

    sample_requests = [
        {"request_id": "req-101", "product": "laptop", "quantity": 1},
        {"request_id": "req-102", "product": "mouse", "quantity": 3},
        {"request_id": "req-101", "product": "laptop", "quantity": 1},
        {"request_id": "req-103", "product": "monitor", "quantity": 0},
        {"request_id": "req-104", "product": "keyboard", "quantity": 2},
    ]

    results = process_order_requests(sample_requests)

    for result in results:
        print(result)
```
