```python
import pandas as pd

# Step 1: Load the CSV into a DataFrame
df = pd.read_csv("students.csv")

# Display the original DataFrame
print("Original DataFrame:")
print(df)
print()

# Step 2: Display shape, column names, and data types
print("Shape of DataFrame:")
print(df.shape)
print()

print("Column Names:")
print(df.columns.tolist())
print()

print("Data Types:")
print(df.dtypes)
print()

# Step 3: Run describe()
print("Describe Output:")
describe_output = df.describe()
print(describe_output)
print()

# Explanation of included/excluded columns
print("Columns included in describe():")
print(describe_output.columns.tolist())
print()

print("Explanation:")
print("By default, df.describe() includes only numeric columns.")
print("Therefore, 'age' and 'score' are included because they are numeric.")
print("The 'name' and 'grade' columns are excluded because they contain string/object data.")
print()

# Step 4: Rename columns without modifying the original DataFrame
renamed_df = df.rename(
    columns={
        "score": "final_score",
        "grade": "letter_grade"
    }
)

print("Renamed DataFrame:")
print(renamed_df)
print()

# Step 5: Select specific columns
selected_columns = renamed_df[["name", "final_score", "letter_grade"]]

print("Selected Columns:")
print(selected_columns)
print()

# Step 6A: Using iloc to access first 3 rows and all columns
print("Using iloc (first 3 rows, all columns):")
print(df.iloc[:3, :])
print()

# Step 6B: Using loc to access rows with index labels 1 and 3
# and only the name and grade columns
print("Using loc (rows 1 and 3, columns name and grade):")
print(df.loc[[1, 3], ["name", "grade"]])
```
