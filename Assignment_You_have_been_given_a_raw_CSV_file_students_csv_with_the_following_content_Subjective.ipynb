```python
import numpy as np

np.random.seed(42)

# ── Task 1: Generate and Inspect the Dataset ──────────────────────────────────
names   = np.random.choice(['Alice', 'Bob', 'Carol', 'David', 'Eva'], size=10)
ages    = np.random.randint(15, 19, size=10)
maths   = np.random.randint(40, 101, size=10)
science = np.random.randint(40, 101, size=10)
english = np.random.randint(40, 101, size=10)

marks = np.array([maths, science, english]).T   # shape (10, 3)

print("Marks array shape :", marks.shape)
print("Marks array dtype :", marks.dtype)

# ── Task 2: Reshape and Slice ─────────────────────────────────────────────────
reshaped = marks.reshape(2, 5, 3)
print("\nReshaped marks array (2, 5, 3):")
print(reshaped)

sliced = marks[:5, :2]
print("\nFirst 5 students — Maths and Science only:")
print(sliced)

# Shape (2, 5, 3) means the 10 student records are divided into 2 groups,
# each group containing 5 students, and each student has 3 subject marks.
# It is like splitting the class into 2 batches of 5 students each.

# ── Task 3: Save Records to File ─────────────────────────────────────────────
with open('student_records.txt', 'w') as f:
    for i in range(10):
        line = f"{names[i]} | {ages[i]} | {maths[i]} | {science[i]} | {english[i]}\n"
        f.write(line)

print("\nRecords written to student_records.txt")

with open('student_records.txt', 'r') as f:
    lines = f.readlines()

print("\nVerification — File Contents:")
for line in lines:
    print(line.strip())
```
