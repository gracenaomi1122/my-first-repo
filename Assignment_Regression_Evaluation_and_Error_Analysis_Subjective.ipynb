```python
import pandas as pd

# -------------------------------
# Define the cross-validation results table
# -------------------------------
cv_results = pd.DataFrame({
    "Alpha": [0, 0.001, 0.01, 0.1, 1, 10, 50, 100, 200, 500, 1000, 5000, 10000],
    "Mean R²": [0.671, 0.671, 0.672, 0.675, 0.682, 0.679, 0.668, 0.651, 0.630, 0.588, 0.541, 0.412, 0.301],
    "Mean MAE": [221500, 221490, 221350, 220800, 219400, 220100, 222800, 226500, 231200, 241000, 252300, 278900, 309400]
})

print("=" * 70)
print("RIDGE REGRESSION CROSS-VALIDATION ANALYSIS")
print("=" * 70)

print("\nCross-Validation Results:")
print(cv_results)

# ---------------------------------------------------
# Section 1: Identify the Optimal Alpha (Sweet Spot)
# ---------------------------------------------------
print("\n" + "=" * 70)
print("1. OPTIMAL ALPHA IDENTIFICATION")
print("=" * 70)

# Highest Mean R² (and acceptable MAE)
best_row = cv_results.loc[cv_results["Mean R²"].idxmax()]

print(f"Optimal Alpha : {best_row['Alpha']}")
print(f"Highest Mean R² : {best_row['Mean R²']:.3f}")
print(f"Mean MAE : {best_row['Mean MAE']:,}")

print("\nConclusion:")
print("Alpha = 1 is the sweet spot because it provides:")
print("- The highest Mean R² (best predictive performance)")
print("- The lowest Mean MAE (smallest prediction error)")
print("- The best balance between model complexity and regularization")

# ---------------------------------------------------
# Section 2: Analyze Alpha Behavior
# ---------------------------------------------------
print("\n" + "=" * 70)
print("2. EFFECT OF ALPHA ON MODEL BEHAVIOR")
print("=" * 70)

print("""
As alpha increases:

• Alpha = 0
  - No regularization.
  - Model is more likely to overfit.

• Alpha from 0.001 to 1
  - Regularization improves generalization.
  - Mean R² increases.
  - Mean MAE decreases.
  - Best performance occurs at Alpha = 1.

• Alpha greater than 1
  - Regularization becomes stronger.
  - Model coefficients shrink excessively.
  - Mean R² steadily decreases.
  - Mean MAE increases.
  - Very large alpha values (5000-10000) cause underfitting.
""")

# ---------------------------------------------------
# Section 3: Final Model Training Strategy
# ---------------------------------------------------
print("=" * 70)
print("3. FINAL MODEL TRAINING & EVALUATION STRATEGY")
print("=" * 70)

print("""
Recommended Strategy:

1. Select Alpha = 1 from cross-validation.
2. Retrain the Ridge Regression model using the entire training dataset.
3. Evaluate the final model on a separate unseen test dataset.
4. Report metrics such as:
   - R²
   - MAE
   - RMSE
5. Deploy the trained model if test performance is satisfactory.
""")

# ---------------------------------------------------
# Section 4: Metric Recommendation
# ---------------------------------------------------
print("=" * 70)
print("4. WHICH ERROR METRIC SHOULD THE BUSINESS USE?")
print("=" * 70)

print("""
Business Goal:
Minimize large vehicle price mispricings.

Recommended Metric:
MSE / RMSE

Justification:
- MAE treats every error equally.
- MSE squares the errors, giving much larger penalties to large mistakes.
- RMSE is the square root of MSE, making it easier to interpret while
  still heavily penalizing large pricing errors.

Therefore, when preventing expensive mispricing is the priority,
RMSE (or MSE) should be preferred over MAE.
""")

# ---------------------------------------------------
# Section 5: Ridge vs Lasso
# ---------------------------------------------------
print("=" * 70)
print("5. RIDGE VS LASSO")
print("=" * 70)

print("""
Ridge Regression:
- Uses L2 regularization.
- Shrinks coefficients toward zero.
- Keeps all features in the model.

Lasso Regression:
- Uses L1 regularization.
- Can reduce some coefficients exactly to zero.
- Performs automatic feature selection.

Key Difference:
Ridge keeps every predictor, while Lasso can completely eliminate
less important predictors.
""")

print("=" * 70)
print("FINAL RECOMMENDATION")
print("=" * 70)
print("Optimal Alpha = 1")
print("Highest Mean R² = 0.682")
print("Lowest Mean MAE = 219,400")
print("Retrain the Ridge model with Alpha = 1 on the full training dataset.")
print("Evaluate on a separate test set using R², MAE, and RMSE.")
```
