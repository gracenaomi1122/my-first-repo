```python
#Task2


import pandas as pd
import numpy as np

# Load Dataset

encodings = ["utf-8", "latin1", "cp1252", "ISO-8859-1"]

for enc in encodings:
    try:
        df = pd.read_csv("ai_learning_lab.csv", encoding=enc)
        print(f"Dataset loaded successfully using '{enc}' encoding.")
        break
    except Exception:
        continue
else:
    print("Error: Unable to load the dataset.")
    exit()

# Remove Duplicate Rows

df.drop_duplicates(inplace=True)


# Remove Extra Spaces

columns_to_strip = ["student_name", "topic", "tool_used"]

for col in columns_to_strip:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip()


# Convert to Title Case

columns_to_title = ["attendance", "lab_completed", "api_used"]

for col in columns_to_title:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip().str.title()


# Convert Numeric Columns

numeric_cols = [
    "assignment_score",
    "quiz_score",
    "study_hours",
    "feedback_rating"
]

for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")


# Replace Invalid Values


# Assignment Score
if "assignment_score" in df.columns:
    df.loc[
        (df["assignment_score"] < 0) |
        (df["assignment_score"] > 100),
        "assignment_score"
    ] = np.nan

# Quiz Score
if "quiz_score" in df.columns:
    df.loc[
        (df["quiz_score"] < 0) |
        (df["quiz_score"] > 100),
        "quiz_score"
    ] = np.nan

# Study Hours
if "study_hours" in df.columns:
    df.loc[
        df["study_hours"] < 0,
        "study_hours"
    ] = np.nan

# Feedback Rating
if "feedback_rating" in df.columns:
    df.loc[
        (df["feedback_rating"] < 1) |
        (df["feedback_rating"] > 5),
        "feedback_rating"
    ] = np.nan


# Fill Missing Values


if "assignment_score" in df.columns:
    df["assignment_score"].fillna(
        df["assignment_score"].mean(),
        inplace=True
    )

if "quiz_score" in df.columns:
    df["quiz_score"].fillna(
        df["quiz_score"].mean(),
        inplace=True
    )

if "study_hours" in df.columns:
    df["study_hours"].fillna(
        df["study_hours"].median(),
        inplace=True
    )

if "feedback_rating" in df.columns:
    df["feedback_rating"].fillna(
        df["feedback_rating"].median(),
        inplace=True
    )


# Print Results

print("\nShape of Cleaned DataFrame:")
print(df.shape)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())


# Save Cleaned Dataset

df.to_csv("cleaned_ai_learning_lab.csv", index=False)

print("\nCleaned dataset saved as 'cleaned_ai_learning_lab.csv'")
```

```python
#Task-3

import pandas as pd
import numpy as np

# ---------------------------------
# Load the Cleaned Dataset
# ---------------------------------
encodings = ["utf-8", "latin1", "cp1252", "ISO-8859-1"]

for enc in encodings:
    try:
        df = pd.read_csv("cleaned_ai_learning_lab.csv", encoding=enc)
        print(f"Dataset loaded successfully using '{enc}' encoding.")
        break
    except Exception:
        continue
else:
    print("Error: Unable to load the cleaned dataset.")
    exit()

# ---------------------------------
# Create total_score Column
# ---------------------------------
df["total_score"] = df["assignment_score"] + df["quiz_score"]

# ---------------------------------
# Create performance_level Column
# ---------------------------------
conditions = [
    df["total_score"] >= 160,
    (df["total_score"] >= 120) & (df["total_score"] < 160),
    (df["total_score"] >= 80) & (df["total_score"] < 120),
    df["total_score"] < 80
]

choices = [
    "Excellent",
    "Good",
    "Average",
    "Needs Support"
]

df["performance_level"] = np.select(conditions, choices, default="Unknown")

# ---------------------------------
# Average Scores
# ---------------------------------
print("\nAverage Assignment Score:")
print(round(df["assignment_score"].mean(), 2))

print("\nAverage Quiz Score:")
print(round(df["quiz_score"].mean(), 2))

print("\nAverage Study Hours:")
print(round(df["study_hours"].mean(), 2))

# ---------------------------------
# Performance Level Count
# ---------------------------------
print("\nStudents in Each Performance Level:")
print(df["performance_level"].value_counts())

# ---------------------------------
# Topic-wise Average Total Score
# ---------------------------------
print("\nAverage Total Score Topic-wise:")
print(df.groupby("topic")["total_score"].mean().round(2))

# ---------------------------------
# Batch-wise Average Total Score
# ---------------------------------
print("\nAverage Total Score Batch-wise:")
print(df.groupby("batch")["total_score"].mean().round(2))

# ---------------------------------
# Students Needing Support
# ---------------------------------
print("\nStudents Who Need Support:")

needs_support = df[df["performance_level"] == "Needs Support"]

if len(needs_support) > 0:
    print(needs_support)
else:
    print("No students need support.")
```

```python
#Task-4


import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import json


# Load the Cleaned Dataset

encodings = ["utf-8", "latin1", "cp1252", "ISO-8859-1"]

for enc in encodings:
    try:
        df = pd.read_csv("cleaned_ai_learning_lab.csv", encoding=enc)
        print(f"Dataset loaded successfully using '{enc}' encoding.")
        break
    except Exception:
        continue
else:
    print("Error: Unable to load the dataset.")
    exit()

# Create total_score Column

df["total_score"] = df["assignment_score"] + df["quiz_score"]


# Create performance_level Column

conditions = [
    df["total_score"] >= 160,
    (df["total_score"] >= 120) & (df["total_score"] < 160),
    (df["total_score"] >= 80) & (df["total_score"] < 120),
    df["total_score"] < 80
]

choices = [
    "Excellent",
    "Good",
    "Average",
    "Needs Support"
]

df["performance_level"] = np.select(conditions, choices, default="Unknown")


# 1. Bar Chart - Average Total Score by Topic
# =================================
topic_score = df.groupby("topic")["total_score"].mean()

plt.figure(figsize=(8, 5))
topic_score.plot(kind="bar")
plt.title("Average Total Score by Topic")
plt.xlabel("Topic")
plt.ylabel("Average Total Score")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("topic_score_chart.png")
plt.close()


# 2. Bar Chart - Student Count by Performance Level
# =================================
performance_count = df["performance_level"].value_counts()

plt.figure(figsize=(6, 5))
performance_count.plot(kind="bar")
plt.title("Student Count by Performance Level")
plt.xlabel("Performance Level")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.savefig("performance_level_chart.png")
plt.close()


# 3. Pie Chart - Attendance Distribution
# =================================
attendance_count = df["attendance"].value_counts()

plt.figure(figsize=(6, 6))
attendance_count.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Attendance Distribution")
plt.ylabel("")
plt.tight_layout()
plt.savefig("attendance_chart.png")
plt.close()

# 4. Line Chart - Average Study Hours by Topic
# =================================
study_hours = df.groupby("topic")["study_hours"].mean()

plt.figure(figsize=(8, 5))
study_hours.plot(kind="line", marker="o")
plt.title("Average Study Hours by Topic")
plt.xlabel("Topic")
plt.ylabel("Average Study Hours")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.savefig("study_hours_chart.png")
plt.close()

# =================================
# Create Summary Dictionary
# =================================
summary = {
    "Total number of students": int(len(df)),
    "Average assignment score": round(df["assignment_score"].mean(), 2),
    "Average quiz score": round(df["quiz_score"].mean(), 2),
    "Average study hours": round(df["study_hours"].mean(), 2),
    "Most common topic": df["topic"].mode()[0]
}

# Save Summary to JSON
with open("learning_summary.json", "w") as file:
    json.dump(summary, file, indent=4)

print("\nVisual reports generated successfully.")
print("Charts saved as:")
print("1. topic_score_chart.png")
print("2. performance_level_chart.png")
print("3. attendance_chart.png")
print("4. study_hours_chart.png")
print("\nSummary saved as learning_summary.json")
```

```python
#Task-5


import pandas as pd
import sqlite3


# Load the Cleaned Dataset

encodings = ["utf-8", "latin1", "cp1252", "ISO-8859-1"]

for enc in encodings:
    try:
        df = pd.read_csv("cleaned_ai_learning_lab.csv", encoding=enc)
        print(f"Dataset loaded successfully using '{enc}' encoding.")
        break
    except Exception:
        continue
else:
    print("Error: Unable to load the dataset.")
    exit()


# Create total_score Column

df["total_score"] = df["assignment_score"] + df["quiz_score"]


# Create SQLite Database

conn = sqlite3.connect("learning_lab.db")

# Store DataFrame into SQLite table
df.to_sql("student_learning", conn, if_exists="replace", index=False)

print("\nData stored successfully in learning_lab.db")


# Query 1
# Average Assignment Score by Topic

print("\n========== Query 1 ==========")
print("Average Assignment Score by Topic")

query1 = """
SELECT topic,
       AVG(assignment_score) AS average_assignment_score
FROM student_learning
GROUP BY topic;
"""

result1 = pd.read_sql_query(query1, conn)
print(result1)

# Query 2
# Student Count by Batch

print("\n========== Query 2 ==========")
print("Student Count by Batch")

query2 = """
SELECT batch,
       COUNT(*) AS student_count
FROM student_learning
GROUP BY batch;
"""

result2 = pd.read_sql_query(query2, conn)
print(result2)

# Query 3
# Students with Total Score Below 80

print("\n========== Query 3 ==========")
print("Students with Total Score Below 80")

query3 = """
SELECT student_name,
       total_score
FROM student_learning
WHERE total_score < 80;
"""

result3 = pd.read_sql_query(query3, conn)
print(result3)


# Close Connection

conn.close()

print("\nSQLite analysis completed successfully.")
```
